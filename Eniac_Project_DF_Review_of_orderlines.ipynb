# Remove the point

```python
#it seems like all the unit price above 1000 have an extra .  after the 1000 denominator.
```

```python
orderlines_df[orderlines_df['unit_price'].astype(str).str.count(r'\.') > 1].tail(10)
```

```python
# removing the thousand denominator's point
orderlines_df['unit_price'] = orderlines_df['unit_price'].astype(str).str.replace(r'\.(?=.*\.)', '', regex=True)
orderlines_df['unit_price']
orderlines_df.sample(10)
```

# Convert Unit_price into numeric and date into date and time

```python
orderlines_df["date"] = pd.to_datetime(orderlines_df["date"])
```

```python
orderlines_df['unit_price'] = pd.to_numeric(orderlines_df['unit_price'])
```
